```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert to datetime
sales['Order_Date'] = pd.to_datetime(sales['Order_Date'], errors='coerce')

# Plot daily total sales
sales.groupby('Order_Date')['Total_Sales'].sum().plot(
    figsize=(10,5),
    title="Daily Total Sales Over Time"
)
plt.ylabel("Total Sales")
plt.xlabel("Order Date")
plt.show()
```

```python
sales.groupby('Order_Date')['Quantity'].sum().plot(
    figsize=(10,5),
    title="Daily Quantity Sold Over Time",
    color='orange'
)
plt.ylabel("Quantity Sold")
plt.xlabel("Order Date")
plt.show()
```

```python
top_products = (
    sales.groupby('Product_ID')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
top_products.plot(kind='bar', title="Top 10 Products by Total Sales", figsize=(8,5))
plt.ylabel("Total Sales")
plt.show()
```

```python
region_sales = sales.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
region_sales.plot(kind='bar', title="Total Sales by Region", figsize=(8,5))
plt.ylabel("Total Sales")
plt.show()
```

```python
discount_region = sales.groupby('Region')['Discount'].mean().sort_values(ascending=False)
discount_region.plot(kind='bar', title="Average Discount by Region", figsize=(8,5))
plt.ylabel("Average Discount")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x='Discount', y='Total_Sales', data=sales, alpha=0.5)
plt.title("Relationship Between Discount and Total Sales")
plt.show()
```

```python
region_sales = sales.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
region_sales.plot(kind='bar', figsize=(8,5), title="Total Sales by Region", color='teal')
plt.ylabel("Total Sales")
plt.show()
```

```python
sales['Month'] = sales['Order_Date'].dt.to_period('M')
monthly_sales = sales.groupby('Month')['Total_Sales'].sum()
monthly_sales.plot(kind='line', figsize=(10,5), title="Monthly Sales Trend", marker='o')
plt.ylabel("Total Sales")
plt.show()
```

```python
top_products = (
    sales.groupby('Product_ID')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
top_products.plot(kind='bar', figsize=(8,5), title="Top 10 Products by Total Sales", color='orange')
plt.ylabel("Total Sales")
plt.show()
```

```python
# You already have the merged dataset; if not, recreate it
merged = pd.merge(inventory, daily_demand, on='Product_ID', how='left')

# Determine if product needs reorder
merged['Need_Reorder'] = merged['Stock_Quantity'] < merged['Reorder_Level']

# Filter products that need reorder
to_reorder = merged[merged['Need_Reorder'] == True]

# Display only key columns
to_reorder[['Product_ID', 'Stock_Quantity', 'Reorder_Level', 'Average_Daily_Demand']]
```

```python
# Load sales data
sales = pd.read_csv('supply_chain_project/data/dataset/ds_sales_data.csv')

# Convert date
sales['Order_Date'] = pd.to_datetime(sales['Order_Date'])

# Calculate daily demand per product
daily_demand = (
    sales.groupby('Product_ID')['Quantity']
    .mean()
    .reset_index()
    .rename(columns={'Quantity': 'Average_Daily_Demand'})
)

daily_demand.head()
```

```python
# Load inventory data
inventory = pd.read_csv('supply_chain_project/data/dataset/ds_inventory_data.csv')

# Merge both datasets
merged = pd.merge(inventory, daily_demand, on='Product_ID', how='left')

merged.head()
```

```python
to_reorder.loc[:, 'Priority_Score'] = to_reorder['Average_Daily_Demand'] / (to_reorder['Stock_Quantity'] + 1)
```

```python
to_reorder = merged[merged['Need_Reorder'] == True].copy()
to_reorder['Priority_Score'] = to_reorder['Average_Daily_Demand'] / (to_reorder['Stock_Quantity'] + 1)
```

```python
to_reorder = to_reorder.sort_values(by='Priority_Score', ascending=False)
to_reorder.head(10)
```

```python
import matplotlib.pyplot as plt

top10 = to_reorder.head(10)
plt.figure(figsize=(10,5))
plt.bar(top10['Product_ID'], top10['Priority_Score'], color='red')
plt.title("Top 10 Products to Reorder Urgently")
plt.xlabel("Product ID")
plt.ylabel("Priority Score (Demand / Stock)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Save main analysis outputs
merged.to_csv("supply_chain_project/data/final_merged_dataset.csv", index=False)
to_reorder.to_csv("supply_chain_project/data/final_reorder_priority_list.csv", index=False)
```

```python
total_products = len(merged)
need_reorder = len(to_reorder)
percent_reorder = round(need_reorder / total_products * 100, 2)

print(f"Total Products: {total_products}")
print(f"Products Needing Reorder: {need_reorder}")
print(f"Percentage Needing Reorder: {percent_reorder}%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.pie([need_reorder, total_products - need_reorder],
        labels=['Need Reorder', 'Sufficient Stock'],
        autopct='%1.1f%%',
        colors=['red', 'green'])
plt.title("Inventory Reorder Status Overview")
plt.show()
```
